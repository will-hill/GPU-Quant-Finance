```python
import sys
import json
import warnings
import numpy as np
import pandas as pd
warnings.filterwarnings("ignore")
sys.path.insert(0, "../gex_history")
import rows as R
import gexlib as G
pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40)
ctx = R.load("spy")
print(ctx["sym"], len(ctx["daily"]), "days")
```

```python
res = R.row10(ctx)
print(json.dumps(R.show(res), indent=1))
_ = R.fig10(ctx, res, G.FIGURES / 'row10_tail_days.png')
```

Every dot is one day's SPY return in percent. The 20 largest moves up or down are enlarged and coloured by the gamma reading at the close before them, orange after a negative close and cyan after a positive close. The note in the corner gives the share of all days that ended with a negative close, which is the split you would expect if the regime made no difference.
